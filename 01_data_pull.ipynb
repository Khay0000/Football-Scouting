```python
import time

leagues = ["EPL", "La_liga", "Bundesliga", "Serie_A", "Ligue_1"]
seasons = ["2022", "2023", "2024"]
num_cols = ["games", "time", "goals", "xG", "assists", "xA", "shots",
            "key_passes", "yellow_cards", "red_cards", "npg", "npxG",
            "xGChain", "xGBuildup"]

frames = []
with UnderstatClient() as understat:
    for league in leagues:
        for season in seasons:
            try:
                data = understat.league(league=league).get_player_data(season=season)
                temp = pd.DataFrame(data)
                temp[num_cols] = temp[num_cols].apply(pd.to_numeric)
                temp["league"] = league
                temp["season"] = season
                temp.to_csv(f"data/raw/understat_{league}_{season}.csv", index=False)
                frames.append(temp)
                print(f"Done: {league} {season} ({len(temp)} players)")
            except Exception as e:
                print(f"Failed: {league} {season} -> {e}")
            time.sleep(2)

all_players = pd.concat(frames, ignore_index=True)
all_players.to_csv("data/raw/understat_all.csv", index=False)
print(all_players.shape)
```

```python
with UnderstatClient() as understat:
    for season in seasons:
        try:
            data = understat.league(league="La_Liga").get_player_data(season=season)
            temp = pd.DataFrame(data)
            temp[num_cols] = temp[num_cols].apply(pd.to_numeric)
            temp["league"] = "La_Liga"
            temp["season"] = season
            temp.to_csv(f"data/raw/understat_La_Liga_{season}.csv", index=False)
            frames.append(temp)
            print(f"Done: La_Liga {season} ({len(temp)} players)")
        except Exception as e:
            print(f"Failed: La_Liga {season} -> {e}")
        time.sleep(2)

all_players = pd.concat(frames, ignore_index=True)
all_players.to_csv("data/raw/understat_all.csv", index=False)
print(all_players.shape)
print(all_players["league"].value_counts())
```
