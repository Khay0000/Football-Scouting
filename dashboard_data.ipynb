```python
import pandas as pd

mids = pd.read_csv("data/processed/midfielders_scored.csv")

cols = ["player_name", "team_title", "league", "season", "age", "market_value",
        "time", "score", "creation_score", "buildup_score", "scoring_score",
        "xA_p90", "key_passes_p90", "xGChain_p90", "xGBuildup_p90",
        "npxG_p90", "shots_p90"]

dash = mids[cols].rename(columns={"time": "minutes", "team_title": "team"}).copy()
dash["league"] = dash["league"].str.replace("_", " ")
dash["market_value_m"] = dash["market_value"] / 1e6
dash["score_per_million"] = dash["score"] / dash["market_value_m"]

dash.to_csv("data/processed/dashboard_data.csv", index=False)
print(dash.shape)
dash.head()
```
